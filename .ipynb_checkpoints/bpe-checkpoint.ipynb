```python
corpus = "the there their then them"
```

```python
# i have to code the byte pair encoding.
# split the text - 1
def split_text(corpus: str) -> str:
    word_list = []
    new_corpus = "".join(corpus.split())
    for word in new_corpus:
        for char in word:
            word_list.append(char)

    return word_list
```

```python
# create the pairs. -2

def pairs(word_list:list[str]) -> list:
    pairs = []
    for char in range(len(word_list) - 1):
        pair = (word_list[char] + word_list[char + 1])
        pairs.append(pair)

    return pairs

```

```python
## ok now our pairs are ready -3

def pair(pairs: list) -> str:
    hashmap = {}
    highest_count = 1 # one pair will be there
    highest_count_pair = []
    for pair in pairs:

        if pair in hashmap:
            hashmap[pair] += 1 # increament value by 1
            if hashmap[pair] > highest_count:
                highest_count = hashmap[pair]
                highest_count_pair.append(pair)

        else:
            hashmap[pair] = 1  # value = 1

    return highest_count_pair[-1] if highest_count_pair else None, highest_count
```

```python
# pairs are ready now, 
# i have to merge word_list (original one..word_list)  -- 4

def merge(pair: str, word_list:list[str]) -> list:

    i = 0
    while i < len(word_list) - 1:
        new_pair = word_list[i] + word_list[i+1]

        if new_pair == pair:
            word_list[i] = new_pair
            word_list.pop(i + 1)
            i += 1
        else:
            i += 1
            
    
    return word_list
```

```python
def bpe_encoding(n_iterations):
    word_list = split_text(corpus) # --1
    for i in range(n_iterations):
        # 1. Changed variable name from 'pairs' to 'current_pairs'
        current_pairs = pairs(word_list) # -- 2
        
        # 2. Changed variable name from 'pair' to 'best_pair'
        best_pair, count = pair(current_pairs) # -- 3
        
        # 3. Passed the updated variable name into the merge function
        word_list = merge(best_pair, word_list) # -- 4
    return word_list, count
```

```python
word_list, count = bpe_encoding(50)
print(word_list)
print(count)
```
