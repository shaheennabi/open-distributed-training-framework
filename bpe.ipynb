```python
corpus = """the there their then them they
these those this that
cat cats cater cattle catch
dog dogs doing done
run runs running runner
walk walks walking walker
play plays playing player
talk talks talking talked
look looks looking looked"""
```

```python
def split_corpus(corpus: str) -> list[list]:

    sequences = corpus.split()
    cleaned_corpus = [list(word) for word in sequences]
    return cleaned_corpus

```

```python
# find the pairs

def make_pairs(cleaned_corpus:list[list]) -> list[list]:

    pairs = [] 
    for i in range(len(cleaned_corpus)):
        j = 0
        lst = []
        while j < len(cleaned_corpus[i]) - 1:
            pair = cleaned_corpus[i][j], cleaned_corpus[i][j + 1]
            lst.append(pair)
            j += 1
        pairs.append(lst)

    return pairs 
```

```python

def pair_frequency_map(pairs: list[list]) -> dict:

    map = {}
    for i in range(len(pairs)):

        j = 0
        while j < len(pairs[i]):
            if pairs[i][j] in map:
                map[pairs[i][j]] += 1
                
            else:
                map[pairs[i][j]] = 1 

            j += 1

    return map
    
```

```python
# count the highest frequency pair ,, for tie-breaking choose first.

def extracting_high_freq_pair(map:dict) -> tuple:
    top_freq_pairs = []
    highest_count = 0

    for key, value in map.items():
        if value > highest_count:
            highest_count = value
            top_freq_pairs = [key]

    return top_freq_pairs[0]
```

```python
# now on the main_corpus, combine the strings

def merge_pairs(highest_frequency_pair: tuple, cleaned_corpus:list[list]) -> list[list]:

    n, m = highest_frequency_pair
    for i in range(len(cleaned_corpus)):
        j = 0
        while j < len(cleaned_corpus[i]) - 1:
            if cleaned_corpus[i][j] == n and cleaned_corpus[i][j+1] == m:
                cleaned_corpus[i][j] = cleaned_corpus[i][j] + cleaned_corpus[i][j+1] # merge here
                cleaned_corpus[i].pop(j+1)

            else:
                j += 1

    return cleaned_corpus
```

```python

def byte_pair_encoding(corpus: str, n_iterations:int) -> list[list]:
    cleaned_corpus = split_corpus(corpus)

    for i in range(n_iterations):
        pairs = make_pairs(cleaned_corpus)
        map = pair_frequency_map(pairs)
        if len(map) == 0:
            break
        highest_frequency_pair = extracting_high_freq_pair(map)
        cleaned_corpus = merge_pairs(highest_frequency_pair, cleaned_corpus)

    return cleaned_corpus
```

```python
final_corpus = byte_pair_encoding(corpus, 10)
print(final_corpus)
```
