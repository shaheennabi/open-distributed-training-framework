```python

def shape_corpus(corpus: str) ->  list[str]:
    cleaned_corpus = corpus.split()
    return cleaned_corpus
```

```python

def byte_corpus(cleaned_corpus:list[str]) -> list[list[int]]:

    byte_corpus = []
    for i in cleaned_corpus:
        encoded = i.encode("utf-8")
        temp_list = []
        for j in encoded:
            temp_list.append(j)
        byte_corpus.append(temp_list)
    return byte_corpus
```

```python
# we need deepseek v3 -- merge rules etc
import requests

url = "https://huggingface.co/deepseek-ai/DeepSeek-V3/resolve/main/tokenizer.json"

response = requests.get(url)
response.raise_for_status()

with open("tokenizer.json", "wb") as f:
    f.write(response.content)

print("Downloaded tokenizer.json")

from pathlib import Path

main_dir = Path.cwd()
tokenizer_path = main_dir / "tokenizer.json"

import json

with open(tokenizer_path, "r", encoding="utf-8") as f:
    tokenizer = json.load(f)
```
